```python
from pathlib import Path

root = Path("gravacoes em cativeiro")

# 1. Renomeia arquivos
for path in sorted(root.rglob("*"), key=lambda p: len(p.parts), reverse=True):
  if path.is_file():
    new_path = path.with_name(path.name.lower())
    if new_path != path and not new_path.exists():
      path.rename(new_path)

# 2. Renomeia diretórios de baixo para cima
dirs = [p for p in root.rglob("*") if p.is_dir()]
for path in sorted(dirs, key=lambda p: len(p.parts), reverse=True):
  new_path = path.with_name(path.name.lower())
  if new_path != path and not new_path.exists():
    path.rename(new_path)
```

```python
from pathlib import Path

import noisereduce as nr
import soundfile as sf

root_dir = Path("gravacoes em cativeiro")
output_dir = Path("gravacoes em cativeiro_noisereduce")
output_dir.mkdir(parents=True, exist_ok=True)

for wav_path in sorted(root_dir.rglob("*.wav")):
  audio, sample_rate = sf.read(wav_path)

  if audio.ndim > 1:
    audio = audio.T
    reduced = nr.reduce_noise(y=audio, sr=sample_rate)
    reduced = reduced.T
  else:
    reduced = nr.reduce_noise(y=audio, sr=sample_rate)

  relative_path = wav_path.relative_to(root_dir)
  destination = output_dir / relative_path
  destination.parent.mkdir(parents=True, exist_ok=True)
  sf.write(destination, reduced, sample_rate)

  print(f"Processado: {wav_path} -> {destination}")
```
